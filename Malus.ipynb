```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df=pd.read_excel("/content/sample_data/Malus.xlsx")
filtrare=df[df['cos a']>=0]
maxI=filtrare['I'].max()
maxIf=filtrare['If'].max()
df['clasificare']=df.apply(lambda rand: 'pozitiv' if rand['cos a']>0 else 'negativ',axis=1)
grupare=df['clasificare'].value_counts()
print(grupare)
meanval=df['cos 2 a'].mean()
print(f"Valoarea medie este {meanval}")
grupareI=df.groupby('clasificare')['I'].sum()
print(grupareI)
meanI=df['I'].mean()
df['Error']=df['I']-meanI
print(df)
sns.set_theme(style="whitegrid")
sns.regplot(data=df,x="cos 2 a",y="If")
plt.title("Regresie")
plt.savefig("raport Malus.png",dpi=380,bbox_inches="tight")
plt.show()
df.to_excel("Malus reconfigurat")

```
