```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_excel("analyse_meteo_records.xlsx")
print(df)
hot_temp=df[df['temperature']>0]
cold_temp=df[df['temperature']<0]
sns.set_theme(style="whitegrid")
sns.histplot(data=hot_temp, x="temperature", kde=True, bins=5, color="purple")
plt.title("analyse real extreme temperatures")
plt.show()
sns.set_theme(style="whitegrid")
sns.histplot(data=cold_temp, x="temperature", kde=True, bins=5, color="aqua")
plt.title("analyse real extreme temperatures")
plt.show()
```
