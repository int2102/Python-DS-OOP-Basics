```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_excel("europeanpeaksgroup.xlsx")
print(df)
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))
sns.lineplot(data=df,x="Mountain_range",y="number_peaks",marker="o")
plt.title("Analyse")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
