```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_excel("space_mission.xlsx")
print(df)
df['Cost_category']=df.apply(lambda row: 'high cost' if row['sum_cost']>200.0 else 'low cost',axis=1)
sns.set_theme(style="whitegrid")
sns.lineplot(data=df,x="company",y="sum_cost",hue="Cost_category",marker="o")
plt.title("Cost Comparison")
plt.show()
```
